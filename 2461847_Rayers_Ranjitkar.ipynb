<br>

```python
# Applying necessary cleaning steps:

# No data types needs to be converted and no removing duplicates required as there are no duplicated rows

# ---------------------------------------------------
# Handling Missing Values:

# First, we checked the number of missing values in each column.
# If only a very small number of values are missing, we drop those rows
# because the data loss is negligible and it avoids making assumptions.

# Then,
# We calculated IQR to check whether outliers exist in numeric columns.
# If outliers exist, median is used since it is less affected by extreme values.
# If outliers do not exist, mean is used.

for each_col in hdi_2022_df.columns:

    # Checking missing values
    null_count = hdi_2022_df[each_col].isnull().sum()

    # Proceeding only if missing values exist
    if null_count > 0:

        # Handling only numeric columns
        if hdi_2022_df[each_col].dtype != 'object':

            print("Column:", each_col)
            print("Missing values:", null_count)

            # If very few missing values, dropping rows
            if null_count <= 4:
                print("Very few missing values → Dropped these null rows")
                hdi_2022_df = hdi_2022_df.dropna(subset=[each_col])

            # Else using IQR-based imputation
            else:
                print("Using IQR to decide mean or median:")

                col_values = hdi_2022_df[each_col].dropna() # removing NaNs temporarily to calculate Q1, Q3 and IQR and filling it later and storing it in temporary df col_values

                Q1 = col_values.quantile(0.25)
                Q3 = col_values.quantile(0.75)
                IQR = Q3 - Q1

                lower_bound = Q1 - 1.5 * IQR
                upper_bound = Q3 + 1.5 * IQR

                outliers = col_values[ (col_values < lower_bound) | (col_values > upper_bound)]

                if len(outliers) > 0:
                    print("Outliers found → Filling with MEDIAN")
                    hdi_2022_df[each_col] = hdi_2022_df[each_col].fillna(col_values.median()) #filling original df
                else:
                    print("No outliers found → Filling with MEAN")
                    hdi_2022_df[each_col] = hdi_2022_df[each_col].fillna(col_values.mean())

            print()
```

```python
print("Missing values after handling:",hdi_2022_df.isnull().sum().sum())
```

```python
# For classifying the countries based on the regions, a dataset was downloaded from world bank site

import pandas as pd

wb_regions_df = pd.read_excel("/content/drive/MyDrive/Dataset/World_Bank_Country_Regions.xlsx")

# Keeping only required columns (not editing wb_regions_df, creating a clean mapping df)
region_map_df = wb_regions_df[["Economy", "Region"]].copy()

# Renaming mapping columns for easier merge
region_map_df = region_map_df.rename(columns={
    "Economy": "country",
    "Region": "region"
})

# Fixing World Bank country names to match HDI dataset country names
wb_name_fix = {
    "Egypt, Arab Rep.": "Egypt",
    "Iran, Islamic Rep.": "Iran",
    "Korea, Rep.": "South Korea",
    "Korea, Dem. People's Rep.": "North Korea",
    "Venezuela, RB": "Venezuela",
    "Syrian Arab Republic": "Syria",
    "Yemen, Rep.": "Yemen",
    "Bahamas, The": "Bahamas",
    "Gambia, The": "Gambia",
    "Congo, Dem. Rep.": "Congo (Democratic Republic of the)",
    "Congo, Rep.": "Congo",
    "Slovak Republic": "Slovakia",
    "Hong Kong SAR, China": "Hong Kong, China (SAR)",
    "West Bank and Gaza": "Palestine, State of",
    "Micronesia, Fed. Sts.": "Micronesia (Federated States of)",
    "Kyrgyz Republic": "Kyrgyzstan",
    "Lao PDR": "Lao People's Democratic Republic",
    "Cote d'Ivoire": "Côte d'Ivoire",
    "Eswatini": "Eswatini (Kingdom of)",
    "Moldova": "Moldova (Republic of)",
    "Tanzania": "Tanzania (United Republic of)",
    "Vietnam": "Viet Nam",
    "Bolivia": "Bolivia (Plurinational State of)",
    "Russian Federation": "Russian Federation",
}

# Applying WB name fixes
region_map_df["country"] = region_map_df["country"].replace(wb_name_fix)

# Selecting only country-level rows from HDI data
country_df = hdi_2020_2022_df[
    hdi_2020_2022_df["iso3"].astype(str).str.len() == 3
].copy()

# Merging region mapping into HDI dataset
country_region_df = pd.merge(
    country_df,
    region_map_df,
    on="country",
    how="left"
)

# Checking missing regions
print("Missing region count:", country_region_df["region"].isnull().sum())
print()


# Dropping rows where region is missing (as the number is negligible)
country_region_df = country_region_df[country_region_df["region"].isnull() == False].copy()


# Grouping by region (country column) and year, and computing mean HDI
avg_hdi_region_year = (
    country_region_df
    .groupby(["region", "year"])["hdi"]
    .mean()
    .reset_index()
)

print(avg_hdi_region_year.head())
print("-" * 60)
```

```python
for year in [2020, 2021, 2022]:
    year_data = avg_hdi_region_year[
        avg_hdi_region_year["year"] == year
    ].sort_values("hdi")

    plt.figure(figsize=(9, 4))
    plt.barh(year_data["region"], year_data["hdi"])

    plt.xlabel("Average HDI", labelpad=15)
    plt.ylabel("Region", labelpad=15)
    plt.title(f"Average HDI by Region ({year})", pad=15)

    plt.xlim(
        avg_hdi_region_year["hdi"].min() - 0.01, #Limiting the value within this range
        avg_hdi_region_year["hdi"].max() + 0.01
    )

    plt.grid(axis="x", linestyle="--", alpha=0.6)
    plt.show()
    print("-" * 80)
```

```python
import pandas as pd

# Checking if hdi_df is loaded/accessible or not
# hdi_df.head()

# Defining South Asia country list

south_asia_countries = [
    "Afghanistan", "Bangladesh", "Bhutan", "India",
    "Maldives", "Nepal", "Pakistan", "Sri Lanka"
]

# Filtering the dataset
hdi_southasia_df = hdi_df[hdi_df["country"].isin(south_asia_countries)].copy()


# Saving the filtered dataset
# hdi_southasia_df.to_csv("HDI_SouthAsia.csv", index=False)
# print("File saved : HDI SouthAsia.csv")
```

**2. Composite Development Score:**

```python
import matplotlib.pyplot as plt

import pandas as pd
import matplotlib.pyplot as plt

# Using latest year (2022) data
southasia_df_2022 = hdi_southasia_df[hdi_southasia_df["year"] == 2022].copy()


# Checking missing values for important columns
print("Missing values check:")
print(southasia_df_2022[["life_expectancy", "gross_inc_percap", "hdi"]].isnull().sum())
print()
print("-"*60)
print()

# Creating Composite Score
southasia_df_2022["composite_score"] = (
    0.30 * southasia_df_2022["life_expectancy"] +
    0.30 * southasia_df_2022["gross_inc_percap"]
)

# Ranking countries by Composite Score
composite_rank_df = southasia_df_2022.sort_values(
    by="composite_score", ascending=False
).reset_index(drop=True)


composite_rank_df["composite_rank"] = composite_rank_df.index + 1

# Ranking countries by HDI
hdi_rank_df = southasia_df_2022.sort_values(
    by="hdi", ascending=False
).reset_index(drop=True)

hdi_rank_df["hdi_rank"] = hdi_rank_df.index + 1

# Comparing rankings
rank_comparison_df = composite_rank_df[[
    "country", "composite_rank", "composite_score", "hdi"
]].copy()

# Adding hdi_rank
hdi_rank_lookup = hdi_rank_df.set_index("country")["hdi_rank"]
rank_comparison_df["hdi_rank"] = hdi_rank_lookup.loc[
    rank_comparison_df["country"]
].values

print("Ranking Comparison:")
print(rank_comparison_df)
print()


# Plotting top 5 countries
top5_df = rank_comparison_df.head(5)
plt.figure(figsize=(8, 4))
plt.bar(top5_df["country"], top5_df["composite_score"])

plt.xlabel("Country", labelpad=10)
plt.ylabel("Composite Score", labelpad=10)
plt.title("Top 5 South Asian Countries by Composite Score (2022)", pad=15)

plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.show()
```

***Comparison of Composite Score Ranking and HDI Ranking***

**- Maldives ranks 1st in Composite Score but 2nd in HDI.**  
This indicates that Maldives performs very strongly in life expectancy and income, which are the two components used in the composite score. However, since HDI also includes education, the slight difference in ranking suggests that other countries may perform marginally better in educational indicators, affecting the overall HDI position.

**- Sri Lanka ranks 2nd in Composite Score but 1st in HDI.**

<br>

**3. Outlier Detection:**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Finding IQR bounds for HDI
q1_hdi = southasia_df_2022["hdi"].quantile(0.25)
q3_hdi = southasia_df_2022["hdi"].quantile(0.75)
iqr_hdi = q3_hdi - q1_hdi

lower_hdi = q1_hdi - 1.5 * iqr_hdi
upper_hdi = q3_hdi + 1.5 * iqr_hdi

# Finding IQR bounds for GNI per Capita
q1_gni = southasia_df_2022["gross_inc_percap"].quantile(0.25)
q3_gni = southasia_df_2022["gross_inc_percap"].quantile(0.75)
iqr_gni = q3_gni - q1_gni

lower_gni = q1_gni - 1.5 * iqr_gni
upper_gni = q3_gni + 1.5 * iqr_gni

# Marking outliers using 1.5 × IQR rule
southasia_df_2022["is_outlier"] = (
    (southasia_df_2022["hdi"] < lower_hdi) | (southasia_df_2022["hdi"] > upper_hdi) |
    (southasia_df_2022["gross_inc_percap"] < lower_gni) | (southasia_df_2022["gross_inc_percap"] > upper_gni)
)

# Printing outlier countries
outliers_df = southasia_df_2022[southasia_df_2022["is_outlier"] == True].copy()

print("Outlier countries (2022):")
if len(outliers_df) == 0:
    print("No outliers found using the 1.5 × IQR rule.")
else:
    print(outliers_df[["country", "hdi", "gross_inc_percap"]])
print()

# Creating scatter plot and highlighting outliers
normal_df = southasia_df_2022[southasia_df_2022["is_outlier"] == False]
outlier_points_df = southasia_df_2022[southasia_df_2022["is_outlier"] == True]

plt.figure(figsize=(8, 5))

plt.scatter(normal_df["gross_inc_percap"], normal_df["hdi"], label="Normal")
plt.scatter(outlier_points_df["gross_inc_percap"], outlier_points_df["hdi"], label="Outlier")

plt.xlabel("GNI per Capita (gross_inc_percap)", labelpad=10)
plt.ylabel("HDI", labelpad=10)
plt.title("GNI per Capita vs HDI (South Asia, 2022)", pad=15)

plt.grid(linestyle="--", alpha=0.6)
plt.legend(bbox_to_anchor=(1, 1)) # Positioning the legend outside the plot i.e. (right(1), top(1) )
plt.show()

```
